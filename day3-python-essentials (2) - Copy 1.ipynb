# 2. Understanding the Print Function

The **print() function** is one of the most fundamental and frequently used functions in Python. **It is used to output data to your screen**. Understanding how to use it effectively is crucial for all Python programmers.

To use the print() function, you need to **pass the text you want to print as an argument**. To do this, you need to wrap the text within either **single quotes (' ')** or **double quotes (" ")**.

```python
print("Hello, word!")
```

### 2.1 Printing Multiple Items

You can print multiple items **by separating them with commas**. Python will insert a **space** between each item by default.

```python
name = "Marvin"
age = 30
city ="Accra"

print("Name:",name, "Age:", age,"City:",city)
```

### 2.2 Printing Format Options

Python provides several ways **to format strings and customize print outputs**.


#### String Concatenation

You can concatenate strings and variables **using the "+" operator**. However, it's important to note that when you concatenate using the "+" operator, **it does not automatically insert a space between each item**.


```python
name

print("Hello, "+ name + "!")
```

#### The str.format() Method

Similar to f-strings, the str.format() method in Python **allows you to insert values into a string**. This method provides an alternative way to format strings by **replacing placeholders with the desired values**.

**NOTE**: It is important to note that the number of placeholders provided in the string using the str.format() method **must match the number of values passed as arguments to the format() method**. If the number of placeholders and values doesn't match, you will encounter an error indicating the mismatch.

```python
name
age

print("My name is {}, {} and I am {} years old".format(name,name, age))
```

#### Use f-Strings to Print Formatted Messages

**f-strings** in Python offer a convenient way to format strings by **embedding variables** and expressions directly within the string itself.

As you gain more experience with Python, you'll find that f-strings are enjoyable and useful in numerous scenarios.

```python
name
age

print(f"My name is {name}, and I am {age} years old")
```

### 2.3 Controlling the End of Line

By default, print adds a **newline character (\n)** at the end. This means that each time you call the print() function, it starts a new line. You can change this behavior **using the end parameter**.

The end parameter allows you to specify a different string to be appended at the end of the printed output instead of the default newline character.

This will output: **Hello World**


```python
print("Hello",end=" ")
print("World")
```

### 2.4 Controlling the Separator

When you print multiple items, you can control the separator between them using the **sep** parameter. By default, this parameter is set to a space character (**sep=" "**). So, if you print multiple items with commas in between them, the sep parameter **automatically inserts a space**.

This will output: **Hello-World**

# 5. Handling Errors & Exceptions

Handling errors and exceptions is an important aspect of writing reliable and robust Python code. During the execution of a program, errors and exceptions can occur, leading to unexpected behavior or program crashes. Therefore, **it is crucial to handle them effectively**.

### 5.1 Understanding Errors and Exceptions

In Python, while errors and exceptions are related concepts and both indicate problems in your code that need to be addressed, they are not exactly the same thing.

#### Errors

Errors in Python generally refer to **syntax errors or mistakes in your code** that prevent it from running at all. These are often detected by the Python interpreter during the parsing phase, before the code is executed. Common examples include syntax errors like missing parentheses or misspelled keywords.

```python
pint("we are done")
```

In the above example, **the closing parenthesis is missing in the print statement**, resulting in a **SyntaxError**. This error occurs because Python expects a closing parenthesis to complete the function call. 

**NOTE**: *Errors are a natural part of computer programming, and regardless of your experience level, you will encounter them in your code. As a skilled programmer, it's important to carefully check the syntax of your code to avoid these errors. By doing so, you can create more robust and user-friendly programs in Python.*

#### Exceptions

Exceptions, on the other hand, occur during the execution of a program **when something unexpected happens** that disrupts the normal flow of the program. This could be due to a variety of reasons such as invalid user input, attempting to divide by zero, or accessing a non-existent file. 

```python
user_input = input("Enter a value: ")
user_input + 5 
```

*Python provides various built-in functions and mechanisms to handle errors and exceptions. By handling these issues properly, you can improve the robustness of your code and prevent it from crashing.*

### 5.2 Common Types of Exceptions

 - **NameError**: This exception is raised when a local or global name is not found. It occurs when you try to access a variable or function that has not been defined or is out of scope.

 - **TypeError**: This exception is raised when an operation or function is applied to an object of an inappropriate type. It occurs when you try to perform an unsupported operation on a data type, such as adding a string and an integer .

 - **ValueError**: This exception is raised when a function receives an argument of the correct type but an inappropriate value. It occurs when you pass an argument to a function that is not within the expected range or does not meet certain conditions .

 - **KeyError**: This exception is raised when you try to access a dictionary key that does not exist.

 - **ZeroDivisionError**: This exception is raised when you attempt to divide by zero, which is an invalid operation in mathematics.

### 5.3 Basic Exception Handling

Exception handling is a mechanism in Python that allows us to handle errors and **prevent our programs from crashing**. It ensures that the program continues executing even when exceptions occur.

#### The try and except Block

The try-except block is used to handle exceptions in Python. The code that might generate an exception is placed inside **the try block**, and the except block is used **to catch and handle the exception**.

```python
# Syntax
try:
    
    # code that may cause an exception
except ExceptionType:
    # code to handle the exception
```

```python
try:
    user_input = input("Enter a value: ")
    user_input + 5
except TypeError:
    print("Error: Cannot add a string to int")
    user_input = int(input("Enter a value: "))
    result=user_input + 5

    
```

```python
result
```

By using the try and except block, **the code can handle the error gracefully** and provide a meaningful error message to the user, **rather than crashing with a traceback**. It allows the program to continue running even if an exception occurs, improving the overall robustness and user experience.

#### Handling Multiple Exceptions

You can define multiple except blocks for different exceptions.


```python
num = int(input("Enter numerator"))
den = int(input("Enter you denominator"))
results = num/den

#ValueError
# ZeroDivisionError 
```

```python
try:
    num = int(input("Enter numerator"))
    den = int(input("Enter you denominator"))
    results = num/den
except ValueError:
    print("Error: Invalid input. please enter numerical values")
    
except ZeroDivisionError:
    print("Error: Division by zero not allowed") 
    
```

#### Catching All Exceptions

You can use a bare **except** to catch all exceptions; however, **this is not recommended for most cases**. By catching specific exceptions, you gain more control over how to handle different types of errors, allowing you to provide targeted solutions or error messages.

```python
try:
    num = int(input("Enter numerator"))
    den = int(input("Enter you denominator"))
    results = num/den
except ValueError:
    print("Error: Invalid input. please enter numerical values")
    
except ZeroDivisionError:
    print("Error: Division by zero not allowed") 

except:
    print("An unexpected error occurred, please reach out to support")
```
